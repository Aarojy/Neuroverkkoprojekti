# Comparison of the CNN and FCN models on CIFAR-10 image dataset.

The objective of this assignment was to construct two neural networks: convolutional neural network (CNN) and the fully connected network (FCN). The models were to be trained using the CIFAR‑10 dataset and their results along with other differences were to be compared. When comparing the the two models, clear differences do indeed emerge.

In terms of performance, the CNN achieves substantially higher test accuracy of 81% compared to the FCN’s modest 54% highlighting its ability to preserve spatial information and learn structural visual features, whereas the FCN struggles with the limitations of a fully flattened input representation

The learning curves further highlight this contrast: the FCN shows a strong tendency to overfit, with a widening gap between training and validation loss, whereas the CNN maintains a far more stable and convergent trajectory thanks to its greater capacity to learn imagebased features and the use of tuned regularization methods.

Parameter counts also differ significantly: the final FCN model contains almost 4 million parameters, whereas the CNN requires only about 550 000. The FCN:s parameter count grows rapidly because the input is fully flattened, while the CNN benefits from shared convolutional filters, resulting in a far more compact and efficient structure. Finally, although the CNN has longer training times per epoch due to the computational cost of convolution operations, it reaches its optimal solution in far fewer epochs than the FCN, making it more efficient overall despite the heavier per epoch workload.

All in all, the results clearly demonstrate, as was predicted, that the CNN is far better suited for the CIFAR‑10 classification task than the FCN. It achieves higher accuracy, shows more stable learning behavior, requires significantly fewer parameters, and converges to an effective solution in fewer epochs despite its heavier computational load per epoch.
